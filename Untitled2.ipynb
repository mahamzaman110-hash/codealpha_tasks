```python
from deep_translator import GoogleTranslator

print("🌍 Language Translation Tool")
print("-" * 40)

text = input("Enter text to translate: ")

source = input("Enter source language code (e.g. en): ")
target = input("Enter target language code (e.g. ur): ")

try:
    translated = GoogleTranslator(
        source=source,
        target=target
    ).translate(text)

    print("\nOriginal Text:", text)
    print("Translated Text:", translated)

except Exception as e:
    print("\nError:", e)
```

```python
import requests
import gradio as gr

def translate_text(text, source, target):
    if not text.strip():
        return "Please enter some text."

    url = "https://api.mymemory.translated.net/get"

    params = {
        "q": text,
        "langpair": f"{source}|{target}"
    }

    try:
        response = requests.get(url, params=params, timeout=10)
        data = response.json()

        if data["responseStatus"] == 200:
            return data["responseData"]["translatedText"]
        else:
            return "Translation error. Please try again."

    except Exception as e:
        return f"Error: {e}"


languages = {
    "English": "en",
    "Urdu": "ur",
    "Spanish": "es",
    "French": "fr",
    "German": "de",
    "Arabic": "ar",
    "Hindi": "hi",
    "Chinese": "zh"
}


def translate_from_dropdown(text, source, target):
    return translate_text(
        text,
        languages[source],
        languages[target]
    )


demo = gr.Interface(
    fn=translate_from_dropdown,
    inputs=[
        gr.Textbox(
            label="Enter Text",
            placeholder="Type your text here..."
        ),
        gr.Dropdown(
            choices=list(languages.keys()),
            value="English",
            label="Source Language"
        ),
        gr.Dropdown(
            choices=list(languages.keys()),
            value="Urdu",
            label="Target Language"
        )
    ],
    outputs=gr.Textbox(
        label="Translated Text"
    ),
    title="🌍 AI Language Translation Tool",
    description="Translate text between multiple languages using a translation API.",
    submit_btn="Translate"
)

demo.launch(share=True)
```

```python
import requests
import gradio as gr

languages = {
    "English": "en",
    "Urdu": "ur",
    "Arabic": "ar",
    "Hindi": "hi",
    "Spanish": "es",
    "French": "fr",
    "German": "de",
    "Chinese": "zh"
}

def translate_text(text, source, target):

    if not text.strip():
        return "Please enter some text."

    url = "https://api.mymemory.translated.net/get"

    params = {
        "q": text,
        "langpair": f"{languages[source]}|{languages[target]}"
    }

    try:
        response = requests.get(url, params=params, timeout=10)
        data = response.json()

        if data.get("responseStatus") == 200:
            return data["responseData"]["translatedText"]

        return "Translation failed. Please try again."

    except Exception as e:
        return f"Error: {e}"


with gr.Blocks(title="AI Language Translation Tool") as demo:

    gr.Markdown(
        """
        # 🌍 AI Language Translation Tool
        ### Translate text quickly between multiple languages
        """
    )

    with gr.Row():

        with gr.Column():
            source = gr.Dropdown(
                choices=list(languages.keys()),
                value="English",
                label="Source Language"
            )

            input_text = gr.Textbox(
                label="Enter Text",
                placeholder="Type your text here...",
                lines=6
            )

        with gr.Column():
            target = gr.Dropdown(
                choices=list(languages.keys()),
                value="Urdu",
                label="Target Language"
            )

            output_text = gr.Textbox(
                label="Translated Text",
                lines=6
            )

    translate_btn = gr.Button(
        "🔄 Translate",
        variant="primary"
    )

    translate_btn.click(
        fn=translate_text,
        inputs=[input_text, source, target],
        outputs=output_text
    )

    gr.Markdown(
        """
        **Powered by Translation API | CodeAlpha AI Internship — Task 1**
        """
    )

demo.launch(share=True)
```
