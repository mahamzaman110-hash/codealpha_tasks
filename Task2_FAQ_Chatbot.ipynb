```python
import re
import nltk
import string

from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

print("Libraries loaded successfully!")
```

```python
faqs = [
    {
        "question": "What is CodeAlpha?",
        "answer": "CodeAlpha is an organization that provides internship and learning opportunities."
    },
    {
        "question": "How can I submit my internship task?",
        "answer": "You can submit your technical task through a GitHub repository link."
    },
    {
        "question": "What is the deadline for task submission?",
        "answer": "The task submission deadline is 30th September 2026."
    },
    {
        "question": "How many tasks do I need to complete?",
        "answer": "You need to complete any 2 or 3 tasks. Additional tasks are optional."
    },
    {
        "question": "Is LinkedIn video posting required?",
        "answer": "No, the LinkedIn video explanation post is optional."
    },
    {
        "question": "How do I get an internship certificate?",
        "answer": "The internship certificate is provided after proper task completion and submission."
    },
    {
        "question": "What should I upload to GitHub?",
        "answer": "Upload your completed technical task files and code to your GitHub repository."
    },
    {
        "question": "Can I complete additional tasks?",
        "answer": "Yes, additional tasks are optional and can be completed for extra learning."
    }
]

questions = [faq["question"] for faq in faqs]
answers = [faq["answer"] for faq in faqs]

print("FAQ dataset created successfully!")
print("Total FAQs:", len(faqs))
```

```python
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return ' '.join(words)

processed_questions = [preprocess_text(q) for q in questions]

print("Text preprocessing completed!")
print("\nOriginal:")
print(questions[0])

print("\nProcessed:")
print(processed_questions[0])
```

```python
vectorizer = TfidfVectorizer()

faq_vectors = vectorizer.fit_transform(processed_questions)

print("TF-IDF vectorization completed!")
print("FAQ vectors shape:", faq_vectors.shape)
```

```python
def get_response(user_question):
    user_question = preprocess_text(user_question)

    user_vector = vectorizer.transform([user_question])

    similarity_scores = cosine_similarity(user_vector, faq_vectors)

    best_match_index = similarity_scores.argmax()
    best_score = similarity_scores[0][best_match_index]

    if best_score < 0.2:
        return "Sorry, I could not find a suitable answer to your question."

    return answers[best_match_index]


print("Chatbot matching function created successfully!")
```

```python
test_questions = [
    "What is CodeAlpha?",
    "How many tasks should I complete?",
    "When is the submission deadline?",
    "Is LinkedIn video required?",
    "What can I upload on GitHub?"
]

for question in test_questions:
    print("You:", question)
    print("Bot:", get_response(question))
    print("-" * 60)
```

```python
print("🤖 CodeAlpha FAQ Chatbot")
print("Type 'exit' to stop the chatbot.")
print("-" * 50)

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Goodbye! 👋")
        break

    response = get_response(user_input)
    print("Bot:", response)
```
